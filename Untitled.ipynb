```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# Análise de Tráfego de Rede com PCA (CICIDS2017)\n",
    "**Autor:** Ivo Rosa (Nº 2024209219)  \n",
    "**Unidade Curricular:** Estatística para as Ciências de Dados  \n",
    "\n",
    "---  \n",
    "## 1. Carregamento, Validação e Limpeza dos Dados\n",
    "\n",
    "O dataset original contém 2 520 751 registos e 53 colunas. Aplica-se a regra de validação de qualidade: remover registos com `Init_Win_bytes_forward == -1`, `Init_Win_bytes_backward == -1` ou `min_seg_size_forward < 0`.\n",
    "\n",
    "**Justificação Estatística:** Os valores `-1` constituem códigos de exceção/invalidação do *CICFlowMeter*. Mantê-los como se fossem medições quantitativas reais distorceria gravemente o cálculo da média e do desvio-padrão durante a estandardização (Z-score), corrompendo a estrutura de covariância da PCA."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "import gc\n",
    "import numpy as np\n",
    "import pandas as pd\n",
    "import matplotlib.pyplot as plt\n",
    "import seaborn as sns\n",
    "from sklearn.preprocessing import StandardScaler\n",
    "from sklearn.decomposition import PCA\n",
    "\n",
    "file_path = \"/kaggle/input/datasets/ericanacletoribeiro/cicids2017-cleaned-and-preprocessed/cicids2017_cleaned.csv\"\n",
    "\n",
    "print(\"A carregar e validar os dados por blocos...\")\n",
    "chunks_clean = []\n",
    "n_original = 0\n",
    "n_removed = 0\n",
    "class_counts_original = {}\n",
    "target_col = None\n",
    "\n",
    "for chunk in pd.read_csv(file_path, chunksize=100_000, low_memory=False):\n",
    "    chunk.columns = chunk.columns.str.strip()\n",
    "    if target_col is None:\n",
    "        target_col = \"Attack Type\" if \"Attack Type\" in chunk.columns else \"Label\"\n",
    "    \n",
    "    n_original += len(chunk)\n",
    "    counts = chunk[target_col].value_counts(dropna=False)\n",
    "    for label, count in counts.items():\n",
    "        class_counts_original[label] = class_counts_original.get(label, 0) + int(count)\n",
    "        \n",
    "    mask_invalid = (\n",
    "        chunk[\"Init_Win_bytes_forward\"].eq(-1) |\n",
    "        chunk[\"Init_Win_bytes_backward\"].eq(-1) |\n",
    "        chunk[\"min_seg_size_forward\"].lt(0)\n",
    "    )\n",
    "    n_removed += int(mask_invalid.sum())\n",
    "    chunks_clean.append(chunk.loc[~mask_invalid].copy())\n",
    "    del chunk, mask_invalid\n",
    "\n",
    "df = pd.concat(chunks_clean, ignore_index=True)\n",
    "del chunks_clean\n",
    "gc.collect()\n",
    "\n",
    "print(f\"Registos originais: {n_original:,}\")\n",
    "print(f\"Registos removidos: {n_removed:,} ({n_removed / n_original * 100:.2f}%)\")\n",
    "print(f\"Registos retidos: {len(df):,}\")"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 2. Impacto da Limpeza na Distribuição das Classes\n",
    "\n",
    "A remoção de 48,22% dos registos não afeta as classes de forma uniforme. A tabela seguinte detalha as taxas de retenção por tipo de tráfego, evidenciando a necessidade de contextualizar os resultados estritamente para a população retida."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "orig_s = pd.Series(class_counts_original, name=\"Registos originais\")\n",
    "ret_s = df[target_col].value_counts(dropna=False).rename(\"Registos retidos\")\n",
    "impact_df = pd.concat([orig_s, ret_s], axis=1).fillna(0).astype(\"int64\")\n",
    "impact_df[\"% original\"] = (impact_df[\"Registos originais\"] / n_original * 100).round(2)\n",
    "impact_df[\"% retidos\"] = (impact_df[\"Registos retidos\"] / len(df) * 100).round(2)\n",
    "impact_df[\"Taxa de retenção (%)\"] = (impact_df[\"Registos retidos\"] / impact_df[\"Registos originais\"] * 100).round(2)\n",
    "display(impact_df.sort_values(\"Registos originais\", ascending=False))"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 3. Seleção de Variáveis e Estandardização (Z-Score)\n",
    "\n",
    "Foram selecionadas **51 variáveis numéricas contínuas**. Exclui-se a variável `Destination Port` por constituir um identificador categórico de rede. A estandardização é realizada via `StandardScaler` (Z-score)."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "num_cols = df.select_dtypes(include=[np.number]).columns.tolist()\n",
    "to_exclude = [c for c in [\"Destination Port\", target_col] if c in num_cols]\n",
    "feature_cols = [c for c in num_cols if c not in to_exclude]\n",
    "\n",
    "X = df[feature_cols].to_numpy(dtype=np.float64)\n",
    "y = df[target_col].to_numpy()\n",
    "\n",
    "scaler = StandardScaler()\n",
    "X_scaled = scaler.fit_transform(X)\n",
    "del X, df\n",
    "gc.collect()\n",
    "print(f\"Matriz da PCA pronta com {X_scaled.shape[1]} variáveis e {X_scaled.shape[0]:,} observações.\")"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 4. Aplicação da PCA e Critérios de Retenção\n",
    "\n",
    "Avaliação dos critérios estatísticos para determinação do número de componentes a reter:\n",
    "1. **Critério de Kaiser ($\lambda > 1$):** Recomenda 13 componentes.\n",
    "2. **Variância Acumulada ($\ge 80\\%$):** 11 componentes cobrem 81,90% da variância.\n",
    "3. **Cotovelo (*Scree Plot*):** Decréscimo acentuado inicial seguido de estabilização gradual."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "pca_full = PCA()\n",
    "pca_full.fit(X_scaled)\n",
    "\n",
    "eigenvalues = pca_full.explained_variance_\n",
    "cum_var = np.cumsum(pca_full.explained_variance_ratio_) * 100\n",
    "\n",
    "print(f\"Componentes com Valor Próprio > 1 (Kaiser): {np.sum(eigenvalues > 1)}\")\n",
    "print(f\"Componentes para 80% de variância: {np.argmax(cum_var >= 80.0) + 1} (explica {cum_var[10]:.2f}%)\")"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 5. Interpretação dos *Loadings* e Projeção dos *Scores*\n",
    "\n",
    "- **PC1 (28,26%):** Associada à variabilidade dos tempos de inatividade (*Idle Max*, *Flow IAT Max*).\n",
    "- **PC2 (10,00%):** Associada à dimensão e variabilidade do tamanho dos pacotes.\n",
    "- **PC3 (8,69%):** Associada à contagem de pacotes e dimensões de cabeçalho."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "pca_3 = PCA(n_components=3)\n",
    "scores_3 = pca_3.fit_transform(X_scaled)\n",
    "\n",
    "# Amostra estratificada para visualização visual de scores\n",
    "sample_idx = []\n",
    "rng = np.random.default_rng(42)\n",
    "for cls in np.unique(y):\n",
    "    idx = np.where(y == cls)[0]\n",
    "    sample_idx.extend(rng.choice(idx, size=min(2000, len(idx)), replace=False))\n",
    "\n",
    "plt.figure(figsize=(10, 6))\n",
    "sns.scatterplot(x=scores_3[sample_idx, 0], y=scores_3[sample_idx, 1], hue=y[sample_idx], alpha=0.5, palette=\"tab10\", s=15)\n",
    "plt.title(\"Projeção dos Scores (PC1 vs PC2) — Amostra Estratificada\")\n",
    "plt.xlabel(f\"PC1 ({pca_3.explained_variance_ratio_[0]*100:.2f}%)\")\n",
    "plt.ylabel(f\"PC2 ({pca_3.explained_variance_ratio_[1]*100:.2f}%)\")\n",
    "plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')\n",
    "plt.tight_layout()\n",
    "plt.show()"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 6. Declaração de Transparência (Uso de Ferramentas IA)\n",
    "\n",
    "*Disclaimer:* Este documento foi revisto e otimizado com o apoio de ferramentas de IA (ChatGPT e Gemini), que auxiliaram na verificação gramatical, correção ortográfica e melhoria da clareza expositiva. Os cálculos estatísticos, gráficos e conclusões derivam estritamente do código executado sobre os dados e são da inteira responsabilidade do autor."
   ]
  }
 ],
 "metadata": {
  "language_info": { "name": "python" }
 },
 "nbformat": 4,
 "nbformat_minor": 5
}
```
